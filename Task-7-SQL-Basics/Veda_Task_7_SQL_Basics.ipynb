```python
import sqlite3
import pandas as pd
import urllib.request
import os

db_url = "https://github.com/lerocha/chinook-database/releases/download/v1.4.5/Chinook_Sqlite.sqlite"
db_file = "chinook.db"

urllib.request.urlretrieve(db_url, db_file)

conn = sqlite3.connect(db_file)

print("Chinook database loaded successfully!")
print("Database file:", db_file)
```

```python
tables = pd.read_sql_query("""
SELECT name
FROM sqlite_master
WHERE type = 'table'
ORDER BY name;
""", conn)

tables
```

## SQL Query 1 – Retrieve Customers

**Question:** What are the first 10 customers in the database?

This query demonstrates the basic `SELECT` statement and `LIMIT`.

```python
query1 = """
SELECT CustomerId, FirstName, LastName, Country
FROM Customer
LIMIT 10;
"""

pd.read_sql_query(query1, conn)
```

## SQL Query 2 – Filter Customers by Country

**Question:** Which customers are from Brazil?

This query demonstrates filtering records using `WHERE`.

```python
query2 = """
SELECT CustomerId, FirstName, LastName, Country
FROM Customer
WHERE Country = 'Brazil';
"""

pd.read_sql_query(query2, conn)
```

## SQL Query 3 – Filter Tracks by Composer

**Question:** Which tracks were composed by a specific composer?

This demonstrates filtering text data using `WHERE`.

```python
query3 = """
SELECT TrackId, Name, Composer
FROM Track
WHERE Composer IS NOT NULL
LIMIT 15;
"""

pd.read_sql_query(query3, conn)
```

## SQL Query 4 – Sort Tracks by Name

**Question:** What are the first 15 tracks in alphabetical order?

This demonstrates sorting using `ORDER BY`.

```python
query4 = """
SELECT TrackId, Name, Composer
FROM Track
ORDER BY Name ASC
LIMIT 15;
"""

pd.read_sql_query(query4, conn)
```

## SQL Query 5 – Highest-Priced Tracks

**Question:** Which tracks have the highest unit prices?

This demonstrates descending sorting.

```python
query5 = """
SELECT TrackId, Name, UnitPrice
FROM Track
ORDER BY UnitPrice DESC
LIMIT 10;
"""

pd.read_sql_query(query5, conn)
```

## SQL Query 6 – Count Customers

**Question:** How many customers are in the database?

This demonstrates the `COUNT()` aggregate function.


```python
query6 = """
SELECT COUNT(*) AS Total_Customers
FROM Customer;
"""

pd.read_sql_query(query6, conn)
```

## SQL Query 7 – Average Track Price

**Question:** What is the average price of a track?

This demonstrates the `AVG()` aggregate function.

```python
query7 = """
SELECT ROUND(AVG(UnitPrice), 2) AS Average_Track_Price
FROM Track;
"""

pd.read_sql_query(query7, conn)
```

## SQL Query 8 – Total Sales

**Question:** What is the total revenue generated from invoices?

This demonstrates the `SUM()` aggregate function.

```python
query8 = """
SELECT ROUND(SUM(Total), 2) AS Total_Revenue
FROM Invoice;
"""

pd.read_sql_query(query8, conn)
```

## SQL Query 9 – Sales by Country

**Question:** How much revenue was generated from each country?

This demonstrates `GROUP BY` with `SUM()`.

```python
query9 = """
SELECT BillingCountry,
       ROUND(SUM(Total), 2) AS Total_Sales
FROM Invoice
GROUP BY BillingCountry
ORDER BY Total_Sales DESC;
"""

pd.read_sql_query(query9, conn)
```

## SQL Query 10 – Number of Customers by Country

**Question:** How many customers belong to each country?

This demonstrates `GROUP BY` with `COUNT()`.

```python
query10 = """
SELECT Country,
       COUNT(*) AS Customer_Count
FROM Customer
GROUP BY Country
ORDER BY Customer_Count DESC;
"""

pd.read_sql_query(query10, conn)
```

## SQL Query 11 – Average Invoice Value by Country

**Question:** What is the average invoice value for each billing country?

```python
query11 = """
SELECT BillingCountry,
       ROUND(AVG(Total), 2) AS Average_Invoice_Value
FROM Invoice
GROUP BY BillingCountry
ORDER BY Average_Invoice_Value DESC;
"""

pd.read_sql_query(query11, conn)
```

## SQL Query 12 – Countries with High Total Sales

**Question:** Which countries generated more than $50 in total sales?

This demonstrates the use of `HAVING` after aggregation.

```python
query12 = """
SELECT BillingCountry,
       ROUND(SUM(Total), 2) AS Total_Sales
FROM Invoice
GROUP BY BillingCountry
HAVING SUM(Total) > 50
ORDER BY Total_Sales DESC;
"""

pd.read_sql_query(query12, conn)
```

## SQL Query 13 – Customers with Multiple Invoices

**Question:** Which customers have more than 5 invoices?

This demonstrates `GROUP BY` and `HAVING`.

```python
query13 = """
SELECT CustomerId,
       COUNT(*) AS Invoice_Count
FROM Invoice
GROUP BY CustomerId
HAVING COUNT(*) > 5
ORDER BY Invoice_Count DESC;
"""

pd.read_sql_query(query13, conn)
```

## SQL Query 14 – Maximum Invoice Amount

**Question:** What is the largest individual invoice amount?

```python
query14 = """
SELECT InvoiceId,
       CustomerId,
       Total
FROM Invoice
ORDER BY Total DESC
LIMIT 10;
"""

pd.read_sql_query(query14, conn)
```

## SQL Query 15 – Minimum and Maximum Track Price

**Question:** What are the minimum and maximum track prices?

```python
query15 = """
SELECT
    MIN(UnitPrice) AS Minimum_Price,
    MAX(UnitPrice) AS Maximum_Price
FROM Track;
"""

pd.read_sql_query(query15, conn)
```

## SQL Query 16 – Number of Tracks by Genre

**Question:** How many tracks are available in each genre?

```python
query16 = """
SELECT g.Name AS Genre,
       COUNT(t.TrackId) AS Track_Count
FROM Genre g
JOIN Track t
ON g.GenreId = t.GenreId
GROUP BY g.GenreId, g.Name
ORDER BY Track_Count DESC;
"""

pd.read_sql_query(query16, conn)
```

## SQL Query 17 – Revenue by Customer

**Question:** Which customers generated the highest total revenue?

```python
query17 = """
SELECT
    c.CustomerId,
    c.FirstName || ' ' || c.LastName AS Customer_Name,
    ROUND(SUM(i.Total), 2) AS Total_Spent
FROM Customer c
JOIN Invoice i
ON c.CustomerId = i.CustomerId
GROUP BY c.CustomerId, c.FirstName, c.LastName
ORDER BY Total_Spent DESC
LIMIT 10;
"""

pd.read_sql_query(query17, conn)
```

## SQL Query 18 – Sales by Employee

**Question:** How much sales revenue is associated with each sales-support employee?

```python
query18 = """
SELECT
    e.EmployeeId,
    e.FirstName || ' ' || e.LastName AS Employee_Name,
    ROUND(SUM(i.Total), 2) AS Sales_Handled
FROM Employee e
JOIN Customer c
ON e.EmployeeId = c.SupportRepId
JOIN Invoice i
ON c.CustomerId = i.CustomerId
GROUP BY e.EmployeeId, e.FirstName, e.LastName
ORDER BY Sales_Handled DESC;
"""

pd.read_sql_query(query18, conn)
```

## SQL Query 19 – Tracks Longer Than 5 Minutes

**Question:** Which tracks have a duration longer than five minutes?

This demonstrates filtering using a numerical condition.

```python
query19 = """
SELECT TrackId,
       Name,
       ROUND(Milliseconds / 60000.0, 2) AS Duration_Minutes
FROM Track
WHERE Milliseconds > 300000
ORDER BY Duration_Minutes DESC
LIMIT 15;
"""

pd.read_sql_query(query19, conn)
```

## SQL Query 20 – Sales by Year

**Question:** How much revenue was generated in each invoice year?

This demonstrates grouping using a date component.

```python
query20 = """
SELECT
    strftime('%Y', InvoiceDate) AS Sales_Year,
    ROUND(SUM(Total), 2) AS Annual_Sales
FROM Invoice
GROUP BY Sales_Year
ORDER BY Sales_Year;
"""

pd.read_sql_query(query20, conn)
```

## SQL Query 21 – Customers from the USA

**Question:** How many customers are from the USA?

```python
query21 = """
SELECT COUNT(*) AS USA_Customers
FROM Customer
WHERE Country = 'USA';
"""

pd.read_sql_query(query21, conn)
```

## SQL Query 22 – Top 10 Most Expensive Tracks

**Question:** Which tracks have the highest prices?

```python
query22 = """
SELECT Name, UnitPrice
FROM Track
ORDER BY UnitPrice DESC
LIMIT 10;
"""

pd.read_sql_query(query22, conn)
```

## SQL Query 23 – Number of Invoices per Country

**Question:** How many invoices were generated in each country?

```python
query23 = """
SELECT BillingCountry,
       COUNT(*) AS Invoice_Count
FROM Invoice
GROUP BY BillingCountry
ORDER BY Invoice_Count DESC;
"""

pd.read_sql_query(query23, conn)
```

## SQL Query 24 – Customers with Spending Above $40

**Question:** Which customers spent more than $40?

```python
query24 = """
SELECT
    c.CustomerId,
    c.FirstName || ' ' || c.LastName AS Customer_Name,
    ROUND(SUM(i.Total), 2) AS Total_Spent
FROM Customer c
JOIN Invoice i
ON c.CustomerId = i.CustomerId
GROUP BY c.CustomerId, c.FirstName, c.LastName
HAVING SUM(i.Total) > 40
ORDER BY Total_Spent DESC;
"""

pd.read_sql_query(query24, conn)
```

## SQL Query 25 – Number of Tracks by Media Type

**Question:** How many tracks exist for each media type?

```python
query25 = """
SELECT
    m.Name AS Media_Type,
    COUNT(t.TrackId) AS Track_Count
FROM MediaType m
JOIN Track t
ON m.MediaTypeId = t.MediaTypeId
GROUP BY m.MediaTypeId, m.Name
ORDER BY Track_Count DESC;
"""

pd.read_sql_query(query25, conn)
```
