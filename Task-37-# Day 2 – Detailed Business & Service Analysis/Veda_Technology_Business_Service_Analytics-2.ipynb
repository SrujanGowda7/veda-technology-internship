```python
# ============================================================
# VEDA TECHNOLOGY BUSINESS & SERVICE ANALYTICS
# DAY 1 - DATASET CREATION, CLEANING & BASIC EDA
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# ------------------------------------------------------------
# 1. CREATE SYNTHETIC DATASET
# ------------------------------------------------------------

np.random.seed(42)

n = 1000

dates = pd.date_range(
    start="2026-01-01",
    end="2026-09-28",
    periods=n
)

source_channels = [
    "LinkedIn",
    "Instagram",
    "Google Search",
    "College Referral",
    "Direct",
    "WhatsApp"
]

business_areas = [
    "Internship",
    "Training",
    "IT Services",
    "Digital Services"
]

services = [
    "Data Science Internship",
    "Web Development Internship",
    "Python Training",
    "Data Analytics Training",
    "Website Development",
    "Mobile App Development",
    "SEO Services",
    "Digital Marketing"
]

customer_types = [
    "Student",
    "Individual",
    "Startup",
    "Business"
]

locations = [
    "Bengaluru",
    "Mysuru",
    "Chennai",
    "Hyderabad",
    "Pune",
    "Other"
]

inquiry_types = [
    "Course Information",
    "Internship Application",
    "Service Pricing",
    "Technical Consultation",
    "General Inquiry"
]

status_options = [
    "New",
    "Contacted",
    "In Progress",
    "Completed",
    "Closed"
]

df = pd.DataFrame({
    "Date": dates,
    "Visitor_ID": [f"VEDA{str(i).zfill(4)}" for i in range(1, n + 1)],
    "Source_Channel": np.random.choice(source_channels, n),
    "Business_Area": np.random.choice(
        business_areas,
        n,
        p=[0.35, 0.25, 0.22, 0.18]
    ),
    "Service": np.random.choice(services, n),
    "Customer_Type": np.random.choice(
        customer_types,
        n,
        p=[0.50, 0.20, 0.15, 0.15]
    ),
    "Location": np.random.choice(locations, n),
    "Inquiry_Type": np.random.choice(inquiry_types, n),
    "Website_Visits": np.random.randint(1, 15, n),
    "Application_Submitted": np.random.choice(
        [0, 1],
        n,
        p=[0.55, 0.45]
    ),
    "Inquiry_Submitted": np.random.choice(
        [0, 1],
        n,
        p=[0.35, 0.65]
    ),
    "Converted": np.random.choice(
        [0, 1],
        n,
        p=[0.72, 0.28]
    ),
    "Revenue": np.random.randint(1000, 50000, n),
    "Response_Time_Hours": np.round(
        np.random.uniform(0.5, 48, n), 2
    ),
    "Satisfaction_Score": np.random.randint(1, 6, n),
    "Status": np.random.choice(status_options, n)
})

# Revenue should only exist for converted customers
df.loc[df["Converted"] == 0, "Revenue"] = 0

print("Dataset created successfully!")
print("Dataset Shape:", df.shape)

print("\nFirst 5 Rows:")
display(df.head())


# ------------------------------------------------------------
# 2. DATASET INFORMATION
# ------------------------------------------------------------

print("\nDATASET INFORMATION")
print("-" * 50)

df.info()

print("\nStatistical Summary:")
display(df.describe())


# ------------------------------------------------------------
# 3. CHECK MISSING VALUES
# ------------------------------------------------------------

print("\nMISSING VALUES")
print("-" * 50)

print(df.isnull().sum())


# ------------------------------------------------------------
# 4. CHECK DUPLICATES
# ------------------------------------------------------------

duplicates = df.duplicated().sum()

print("\nDuplicate Rows:", duplicates)

if duplicates > 0:
    df = df.drop_duplicates()

print("Shape after removing duplicates:", df.shape)


# ------------------------------------------------------------
# 5. DATA TYPE CHECK
# ------------------------------------------------------------

df["Date"] = pd.to_datetime(df["Date"])

print("\nData Types:")
print(df.dtypes)


# ------------------------------------------------------------
# 6. BASIC BUSINESS KPIs
# ------------------------------------------------------------

total_visitors = df["Visitor_ID"].nunique()

total_visits = df["Website_Visits"].sum()

total_applications = df["Application_Submitted"].sum()

total_inquiries = df["Inquiry_Submitted"].sum()

total_conversions = df["Converted"].sum()

total_revenue = df["Revenue"].sum()

conversion_rate = (
    total_conversions / total_visitors
) * 100

average_satisfaction = df["Satisfaction_Score"].mean()

average_response_time = df["Response_Time_Hours"].mean()


print("\n" + "=" * 55)
print("VEDA TECHNOLOGY - BASIC BUSINESS KPIs")
print("=" * 55)

print("Total Unique Visitors:", total_visitors)
print("Total Website Visits:", total_visits)
print("Total Applications:", total_applications)
print("Total Inquiries:", total_inquiries)
print("Total Conversions:", total_conversions)
print("Conversion Rate: {:.2f}%".format(conversion_rate))
print("Total Revenue: ₹{:,.0f}".format(total_revenue))
print(
    "Average Satisfaction Score: {:.2f}/5".format(
        average_satisfaction
    )
)
print(
    "Average Response Time: {:.2f} hours".format(
        average_response_time
    )
)


# ------------------------------------------------------------
# 7. BUSINESS AREA ANALYSIS
# ------------------------------------------------------------

print("\nBUSINESS AREA DISTRIBUTION")
print("-" * 50)

business_area_counts = df["Business_Area"].value_counts()

print(business_area_counts)

plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="Business_Area",
    order=df["Business_Area"].value_counts().index
)

plt.title("Business Area Demand")
plt.xlabel("Business Area")
plt.ylabel("Number of Records")
plt.xticks(rotation=20)

plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 8. MARKETING CHANNEL ANALYSIS
# ------------------------------------------------------------

print("\nMARKETING CHANNEL DISTRIBUTION")
print("-" * 50)

channel_counts = df["Source_Channel"].value_counts()

print(channel_counts)

plt.figure(figsize=(9, 5))

sns.countplot(
    data=df,
    x="Source_Channel",
    order=df["Source_Channel"].value_counts().index
)

plt.title("Visitors by Marketing Channel")
plt.xlabel("Marketing Channel")
plt.ylabel("Number of Visitors")
plt.xticks(rotation=30)

plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 9. SERVICE DEMAND ANALYSIS
# ------------------------------------------------------------

print("\nSERVICE DEMAND")
print("-" * 50)

service_demand = df["Service"].value_counts()

print(service_demand)

plt.figure(figsize=(10, 6))

sns.countplot(
    data=df,
    y="Service",
    order=df["Service"].value_counts().index
)

plt.title("Demand by Service")
plt.xlabel("Number of Requests / Visitors")
plt.ylabel("Service")

plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 10. CUSTOMER TYPE ANALYSIS
# ------------------------------------------------------------

print("\nCUSTOMER TYPE DISTRIBUTION")
print("-" * 50)

print(df["Customer_Type"].value_counts())

plt.figure(figsize=(7, 5))

sns.countplot(
    data=df,
    x="Customer_Type",
    order=df["Customer_Type"].value_counts().index
)

plt.title("Customer Type Distribution")
plt.xlabel("Customer Type")
plt.ylabel("Count")

plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 11. LOCATION ANALYSIS
# ------------------------------------------------------------

print("\nLOCATION DISTRIBUTION")
print("-" * 50)

print(df["Location"].value_counts())

plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="Location",
    order=df["Location"].value_counts().index
)

plt.title("Visitors / Customers by Location")
plt.xlabel("Location")
plt.ylabel("Count")
plt.xticks(rotation=20)

plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 12. CONVERSION BY MARKETING CHANNEL
# ------------------------------------------------------------

channel_conversion = (
    df.groupby("Source_Channel")["Converted"]
    .mean()
    .mul(100)
    .sort_values(ascending=False)
)

print("\nCONVERSION RATE BY MARKETING CHANNEL (%)")
print("-" * 50)

print(channel_conversion.round(2))

plt.figure(figsize=(9, 5))

channel_conversion.plot(kind="bar")

plt.title("Conversion Rate by Marketing Channel")
plt.xlabel("Marketing Channel")
plt.ylabel("Conversion Rate (%)")
plt.xticks(rotation=30)

plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 13. REVENUE BY BUSINESS AREA
# ------------------------------------------------------------

revenue_area = (
    df.groupby("Business_Area")["Revenue"]
    .sum()
    .sort_values(ascending=False)
)

print("\nREVENUE BY BUSINESS AREA")
print("-" * 50)

print(revenue_area)

plt.figure(figsize=(8, 5))

revenue_area.plot(kind="bar")

plt.title("Revenue by Business Area")
plt.xlabel("Business Area")
plt.ylabel("Revenue (₹)")
plt.xticks(rotation=20)

plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 14. MONTHLY BUSINESS ACTIVITY
# ------------------------------------------------------------

df["Month"] = df["Date"].dt.to_period("M").astype(str)

monthly_activity = (
    df.groupby("Month")
    .size()
)

print("\nMONTHLY BUSINESS ACTIVITY")
print("-" * 50)

print(monthly_activity)

plt.figure(figsize=(10, 5))

monthly_activity.plot(
    marker="o"
)

plt.title("Monthly Business Activity")
plt.xlabel("Month")
plt.ylabel("Number of Records")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 15. CORRELATION ANALYSIS
# ------------------------------------------------------------

numeric_columns = [
    "Website_Visits",
    "Application_Submitted",
    "Inquiry_Submitted",
    "Converted",
    "Revenue",
    "Response_Time_Hours",
    "Satisfaction_Score"
]

correlation = df[numeric_columns].corr()

print("\nCORRELATION MATRIX")
display(correlation.round(2))

plt.figure(figsize=(9, 6))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Business Metrics Correlation")

plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 16. EXPORT CLEANED DATASET
# ------------------------------------------------------------

file_name = "veda_business_service_analytics_cleaned.csv"

df.to_csv(
    file_name,
    index=False
)

print("\n" + "=" * 55)
print("DAY 1 COMPLETED SUCCESSFULLY")
print("=" * 55)

print("Final Dataset Shape:", df.shape)
print("Cleaned dataset saved as:", file_name)
```

```python
# ============================================================
# VEDA TECHNOLOGY BUSINESS & SERVICE ANALYTICS
# DAY 2 - DETAILED BUSINESS & SERVICE ANALYSIS
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("=" * 65)
print("DAY 2 - DETAILED BUSINESS & SERVICE ANALYSIS")
print("=" * 65)

# ------------------------------------------------------------
# 1. VERIFY DATASET
# ------------------------------------------------------------

print("\nDataset Shape:", df.shape)
print("\nFirst 5 Records:")
display(df.head())

print("\nColumns:")
print(df.columns.tolist())


# ------------------------------------------------------------
# 2. WEBSITE VISITOR ANALYSIS
# ------------------------------------------------------------

print("\n" + "=" * 65)
print("1. WEBSITE VISITOR ANALYSIS")
print("=" * 65)

total_visits = df["Website_Visits"].sum()
avg_visits = df["Website_Visits"].mean()

print("Total Website Visits:", total_visits)
print("Average Visits per Visitor:", round(avg_visits, 2))

channel_visits = (
    df.groupby("Source_Channel")["Website_Visits"]
    .sum()
    .sort_values(ascending=False)
)

print("\nWebsite Visits by Source Channel:")
print(channel_visits)

plt.figure(figsize=(9, 5))
channel_visits.plot(kind="bar")

plt.title("Website Visits by Source Channel")
plt.xlabel("Source Channel")
plt.ylabel("Website Visits")
plt.xticks(rotation=30)

plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 3. INTERNSHIP APPLICATION ANALYSIS
# ------------------------------------------------------------

print("\n" + "=" * 65)
print("2. INTERNSHIP APPLICATION ANALYSIS")
print("=" * 65)

internship_df = df[df["Business_Area"] == "Internship"]

internship_records = len(internship_df)
internship_applications = internship_df["Application_Submitted"].sum()

print("Total Internship Records:", internship_records)
print("Internship Applications:", internship_applications)

if internship_records > 0:
    internship_application_rate = (
        internship_applications / internship_records
    ) * 100

    print(
        "Internship Application Rate:",
        round(internship_application_rate, 2),
        "%"
    )

internship_source = (
    internship_df.groupby("Source_Channel")["Application_Submitted"]
    .sum()
    .sort_values(ascending=False)
)

print("\nInternship Applications by Source:")
print(internship_source)

plt.figure(figsize=(9, 5))
internship_source.plot(kind="bar")

plt.title("Internship Applications by Source Channel")
plt.xlabel("Source Channel")
plt.ylabel("Applications")
plt.xticks(rotation=30)

plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 4. TRAINING PROGRAM ANALYSIS
# ------------------------------------------------------------

print("\n" + "=" * 65)
print("3. TRAINING PROGRAM PERFORMANCE")
print("=" * 65)

training_df = df[df["Business_Area"] == "Training"]

print("Total Training Records:", len(training_df))

training_services = (
    training_df["Service"]
    .value_counts()
)

print("\nTraining Service Demand:")
print(training_services)

training_conversions = (
    training_df.groupby("Service")["Converted"]
    .agg(["count", "sum", "mean"])
)

training_conversions["Conversion_Rate_%"] = (
    training_conversions["mean"] * 100
)

training_conversions = training_conversions.sort_values(
    "Conversion_Rate_%",
    ascending=False
)

print("\nTraining Service Performance:")
print(
    training_conversions[
        ["count", "sum", "Conversion_Rate_%"]
    ].round(2)
)

plt.figure(figsize=(10, 5))

training_services.plot(kind="bar")

plt.title("Training Program Demand")
plt.xlabel("Training / Service")
plt.ylabel("Number of Records")
plt.xticks(rotation=45, ha="right")

plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 5. IT SERVICE INQUIRY ANALYSIS
# ------------------------------------------------------------

print("\n" + "=" * 65)
print("4. IT SERVICE INQUIRY ANALYSIS")
print("=" * 65)

it_df = df[df["Business_Area"] == "IT Services"]

print("Total IT Service Records:", len(it_df))
print(
    "Total IT Service Inquiries:",
    it_df["Inquiry_Submitted"].sum()
)

it_service_demand = (
    it_df["Service"]
    .value_counts()
)

print("\nIT Service Demand:")
print(it_service_demand)

plt.figure(figsize=(10, 5))

it_service_demand.plot(kind="bar")

plt.title("IT Service Demand")
plt.xlabel("Service")
plt.ylabel("Number of Records")
plt.xticks(rotation=45, ha="right")

plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 6. DIGITAL SERVICE ANALYSIS
# ------------------------------------------------------------

print("\n" + "=" * 65)
print("5. DIGITAL SERVICE INQUIRY ANALYSIS")
print("=" * 65)

digital_df = df[df["Business_Area"] == "Digital Services"]

print("Total Digital Service Records:", len(digital_df))
print(
    "Digital Service Inquiries:",
    digital_df["Inquiry_Submitted"].sum()
)

digital_service_demand = (
    digital_df["Service"]
    .value_counts()
)

print("\nDigital Service Demand:")
print(digital_service_demand)

plt.figure(figsize=(10, 5))

digital_service_demand.plot(kind="bar")

plt.title("Digital Service Demand")
plt.xlabel("Service")
plt.ylabel("Number of Records")
plt.xticks(rotation=45, ha="right")

plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 7. INQUIRY TYPE ANALYSIS
# ------------------------------------------------------------

print("\n" + "=" * 65)
print("6. CUSTOMER / SERVICE REQUEST ANALYSIS")
print("=" * 65)

inquiry_analysis = (
    df["Inquiry_Type"]
    .value_counts()
)

print("Inquiry Type Distribution:")
print(inquiry_analysis)

plt.figure(figsize=(10, 5))

inquiry_analysis.plot(kind="bar")

plt.title("Customer Inquiry Type Distribution")
plt.xlabel("Inquiry Type")
plt.ylabel("Number of Records")
plt.xticks(rotation=35, ha="right")

plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 8. INQUIRIES BY CUSTOMER TYPE
# ------------------------------------------------------------

customer_inquiries = (
    df.groupby("Customer_Type")["Inquiry_Submitted"]
    .sum()
    .sort_values(ascending=False)
)

print("\nInquiries by Customer Type:")
print(customer_inquiries)

plt.figure(figsize=(8, 5))

customer_inquiries.plot(kind="bar")

plt.title("Inquiries by Customer Type")
plt.xlabel("Customer Type")
plt.ylabel("Total Inquiries")
plt.xticks(rotation=20)

plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 9. MARKETING CHANNEL PERFORMANCE
# ------------------------------------------------------------

print("\n" + "=" * 65)
print("7. MARKETING CHANNEL PERFORMANCE")
print("=" * 65)

marketing = (
    df.groupby("Source_Channel")
    .agg(
        Visitors=("Visitor_ID", "count"),
        Website_Visits=("Website_Visits", "sum"),
        Applications=("Application_Submitted", "sum"),
        Inquiries=("Inquiry_Submitted", "sum"),
        Conversions=("Converted", "sum"),
        Revenue=("Revenue", "sum")
    )
)

marketing["Conversion_Rate_%"] = (
    marketing["Conversions"] /
    marketing["Visitors"] * 100
)

marketing = marketing.sort_values(
    "Conversion_Rate_%",
    ascending=False
)

print("\nMarketing Channel Performance:")
display(marketing.round(2))


# ------------------------------------------------------------
# 10. MARKETING CONVERSION VISUALIZATION
# ------------------------------------------------------------

plt.figure(figsize=(9, 5))

marketing["Conversion_Rate_%"].plot(kind="bar")

plt.title("Marketing Channel Conversion Rate")
plt.xlabel("Marketing Channel")
plt.ylabel("Conversion Rate (%)")
plt.xticks(rotation=30)

plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 11. REVENUE BY MARKETING CHANNEL
# ------------------------------------------------------------

plt.figure(figsize=(9, 5))

marketing["Revenue"].sort_values(
    ascending=False
).plot(kind="bar")

plt.title("Revenue by Marketing Channel")
plt.xlabel("Marketing Channel")
plt.ylabel("Revenue (₹)")
plt.xticks(rotation=30)

plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 12. BUSINESS AREA PERFORMANCE
# ------------------------------------------------------------

print("\n" + "=" * 65)
print("8. BUSINESS AREA PERFORMANCE")
print("=" * 65)

business_performance = (
    df.groupby("Business_Area")
    .agg(
        Records=("Visitor_ID", "count"),
        Applications=("Application_Submitted", "sum"),
        Inquiries=("Inquiry_Submitted", "sum"),
        Conversions=("Converted", "sum"),
        Revenue=("Revenue", "sum"),
        Avg_Satisfaction=("Satisfaction_Score", "mean"),
        Avg_Response_Time=("Response_Time_Hours", "mean")
    )
)

business_performance["Conversion_Rate_%"] = (
    business_performance["Conversions"] /
    business_performance["Records"] * 100
)

business_performance = business_performance.sort_values(
    "Revenue",
    ascending=False
)

print("\nBusiness Area Performance:")
display(business_performance.round(2))


# ------------------------------------------------------------
# 13. SERVICE DEMAND ANALYSIS
# ------------------------------------------------------------

print("\n" + "=" * 65)
print("9. SERVICE DEMAND ANALYSIS")
print("=" * 65)

service_performance = (
    df.groupby("Service")
    .agg(
        Demand=("Visitor_ID", "count"),
        Inquiries=("Inquiry_Submitted", "sum"),
        Conversions=("Converted", "sum"),
        Revenue=("Revenue", "sum")
    )
)

service_performance["Conversion_Rate_%"] = (
    service_performance["Conversions"] /
    service_performance["Demand"] * 100
)

service_performance = service_performance.sort_values(
    "Demand",
    ascending=False
)

print("\nService Performance:")
display(service_performance.round(2))


# ------------------------------------------------------------
# 14. SERVICE DEMAND VISUALIZATION
# ------------------------------------------------------------

plt.figure(figsize=(10, 6))

service_performance["Demand"].plot(
    kind="barh"
)

plt.title("Overall Service Demand")
plt.xlabel("Demand")
plt.ylabel("Service")

plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 15. LOCATION ANALYSIS
# ------------------------------------------------------------

print("\n" + "=" * 65)
print("10. LOCATION PERFORMANCE")
print("=" * 65)

location_performance = (
    df.groupby("Location")
    .agg(
        Visitors=("Visitor_ID", "count"),
        Inquiries=("Inquiry_Submitted", "sum"),
        Conversions=("Converted", "sum"),
        Revenue=("Revenue", "sum")
    )
)

location_performance["Conversion_Rate_%"] = (
    location_performance["Conversions"] /
    location_performance["Visitors"] * 100
)

location_performance = location_performance.sort_values(
    "Conversions",
    ascending=False
)

display(location_performance.round(2))


# ------------------------------------------------------------
# 16. CUSTOMER TYPE PERFORMANCE
# ------------------------------------------------------------

print("\n" + "=" * 65)
print("11. CUSTOMER TYPE PERFORMANCE")
print("=" * 65)

customer_performance = (
    df.groupby("Customer_Type")
    .agg(
        Customers=("Visitor_ID", "count"),
        Applications=("Application_Submitted", "sum"),
        Inquiries=("Inquiry_Submitted", "sum"),
        Conversions=("Converted", "sum"),
        Revenue=("Revenue", "sum"),
        Avg_Satisfaction=("Satisfaction_Score", "mean")
    )
)

customer_performance["Conversion_Rate_%"] = (
    customer_performance["Conversions"] /
    customer_performance["Customers"] * 100
)

display(
    customer_performance
    .sort_values("Revenue", ascending=False)
    .round(2)
)


# ------------------------------------------------------------
# 17. RESPONSE TIME ANALYSIS
# ------------------------------------------------------------

print("\n" + "=" * 65)
print("12. RESPONSE TIME ANALYSIS")
print("=" * 65)

response_by_area = (
    df.groupby("Business_Area")["Response_Time_Hours"]
    .mean()
    .sort_values()
)

print("\nAverage Response Time by Business Area:")
print(response_by_area.round(2))

plt.figure(figsize=(8, 5))

response_by_area.plot(kind="bar")

plt.title("Average Response Time by Business Area")
plt.xlabel("Business Area")
plt.ylabel("Average Response Time (Hours)")
plt.xticks(rotation=20)

plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 18. SATISFACTION ANALYSIS
# ------------------------------------------------------------

print("\n" + "=" * 65)
print("13. CUSTOMER SATISFACTION ANALYSIS")
print("=" * 65)

satisfaction_area = (
    df.groupby("Business_Area")["Satisfaction_Score"]
    .mean()
    .sort_values(ascending=False)
)

print("\nAverage Satisfaction by Business Area:")
print(satisfaction_area.round(2))

plt.figure(figsize=(8, 5))

satisfaction_area.plot(kind="bar")

plt.title("Average Satisfaction by Business Area")
plt.xlabel("Business Area")
plt.ylabel("Average Satisfaction Score")
plt.xticks(rotation=20)

plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 19. MONTHLY PERFORMANCE ANALYSIS
# ------------------------------------------------------------

print("\n" + "=" * 65)
print("14. MONTHLY BUSINESS PERFORMANCE")
print("=" * 65)

# Ensure Month exists
df["Month"] = df["Date"].dt.to_period("M").astype(str)

monthly_performance = (
    df.groupby("Month")
    .agg(
        Records=("Visitor_ID", "count"),
        Applications=("Application_Submitted", "sum"),
        Inquiries=("Inquiry_Submitted", "sum"),
        Conversions=("Converted", "sum"),
        Revenue=("Revenue", "sum")
    )
)

monthly_performance["Conversion_Rate_%"] = (
    monthly_performance["Conversions"] /
    monthly_performance["Records"] * 100
)

print("\nMonthly Performance:")
display(monthly_performance.round(2))


# ------------------------------------------------------------
# 20. MONTHLY REVENUE TREND
# ------------------------------------------------------------

plt.figure(figsize=(10, 5))

monthly_performance["Revenue"].plot(
    marker="o"
)

plt.title("Monthly Revenue Trend")
plt.xlabel("Month")
plt.ylabel("Revenue (₹)")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 21. DAY 2 KPI SUMMARY
# ------------------------------------------------------------

print("\n" + "=" * 65)
print("DAY 2 - BUSINESS KPI SUMMARY")
print("=" * 65)

print("Total Website Visits:", df["Website_Visits"].sum())
print("Total Applications:", df["Application_Submitted"].sum())
print("Total Inquiries:", df["Inquiry_Submitted"].sum())
print("Total Conversions:", df["Converted"].sum())

conversion_rate = (
    df["Converted"].sum() /
    len(df) * 100
)

print(
    "Overall Conversion Rate:",
    round(conversion_rate, 2),
    "%"
)

print(
    "Total Revenue: ₹{:,.0f}".format(
        df["Revenue"].sum()
    )
)

print(
    "Average Satisfaction:",
    round(df["Satisfaction_Score"].mean(), 2),
    "/ 5"
)

print(
    "Average Response Time:",
    round(df["Response_Time_Hours"].mean(), 2),
    "hours"
)


# ------------------------------------------------------------
# 22. AUTOMATIC TOP-PERFORMER FINDINGS
# ------------------------------------------------------------

top_channel = marketing["Conversion_Rate_%"].idxmax()
top_channel_rate = marketing["Conversion_Rate_%"].max()

top_service = service_performance["Demand"].idxmax()
top_service_demand = service_performance["Demand"].max()

top_revenue_service = service_performance["Revenue"].idxmax()
top_revenue = service_performance["Revenue"].max()

top_business = business_performance["Revenue"].idxmax()

top_location = location_performance["Conversions"].idxmax()

print("\n" + "=" * 65)
print("KEY BUSINESS FINDINGS")
print("=" * 65)

print(
    f"1. Highest converting marketing channel: "
    f"{top_channel} ({top_channel_rate:.2f}%)"
)

print(
    f"2. Highest demand service: "
    f"{top_service} ({top_service_demand} records)"
)

print(
    f"3. Highest revenue service: "
    f"{top_revenue_service} "
    f"(₹{top_revenue:,.0f})"
)

print(
    f"4. Highest revenue business area: "
    f"{top_business}"
)

print(
    f"5. Location with highest conversions: "
    f"{top_location}"
)


# ------------------------------------------------------------
# 23. EXPORT DAY 2 SUMMARY TABLES
# ------------------------------------------------------------

marketing.to_csv(
    "day2_marketing_channel_performance.csv"
)

business_performance.to_csv(
    "day2_business_area_performance.csv"
)

service_performance.to_csv(
    "day2_service_performance.csv"
)

location_performance.to_csv(
    "day2_location_performance.csv"
)

customer_performance.to_csv(
    "day2_customer_performance.csv"
)

monthly_performance.to_csv(
    "day2_monthly_performance.csv"
)

print("\n" + "=" * 65)
print("DAY 2 COMPLETED SUCCESSFULLY")
print("=" * 65)

print("\nFiles Created:")
print("1. day2_marketing_channel_performance.csv")
print("2. day2_business_area_performance.csv")
print("3. day2_service_performance.csv")
print("4. day2_location_performance.csv")
print("5. day2_customer_performance.csv")
print("6. day2_monthly_performance.csv")
```
