```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully!")
```

```python
df = pd.read_csv("/content/titanic_cleaned.csv")

print("Cleaned Titanic dataset loaded successfully!")
print("Shape:", df.shape)

df.head()
```

```python
print("Dataset Shape:", df.shape)

print("\nData Types:")
print(df.dtypes)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nSummary Statistics:")
print(df.describe())
```

```python
plt.figure(figsize=(6, 4))

sns.countplot(x="Survived", data=df)

plt.title("Passenger Survival Count")
plt.xlabel("Survived (0 = No, 1 = Yes)")
plt.ylabel("Number of Passengers")

plt.show()
```

```python
plt.figure(figsize=(7, 5))

sns.countplot(x="Sex", hue="Survived", data=df)

plt.title("Survival Count by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Passengers")
plt.legend(title="Survived", labels=["No", "Yes"])

plt.show()
```

```python
survival_by_sex = df.groupby("Sex")["Survived"].mean() * 100

print("Survival Rate by Gender:")
print(survival_by_sex)
```

```python
plt.figure(figsize=(7, 5))

sns.countplot(x="Pclass", hue="Survived", data=df)

plt.title("Survival Count by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Number of Passengers")
plt.legend(title="Survived", labels=["No", "Yes"])

plt.show()
```

```python
survival_by_class = df.groupby("Pclass")["Survived"].mean() * 100

print("Survival Rate by Passenger Class:")
print(survival_by_class)
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(df["Age"], bins=20, kde=True)

plt.title("Age Distribution of Titanic Passengers")
plt.xlabel("Age")
plt.ylabel("Number of Passengers")

plt.show()
```

```python
plt.figure(figsize=(7, 4))

sns.boxplot(x=df["Age"])

plt.title("Age Distribution and Outliers")
plt.xlabel("Age")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(df["Fare"], bins=30, kde=True)

plt.title("Fare Distribution of Titanic Passengers")
plt.xlabel("Fare")
plt.ylabel("Number of Passengers")

plt.show()
```

```python
plt.figure(figsize=(8, 4))

sns.boxplot(x=df["Fare"])

plt.title("Fare Distribution and Potential Outliers")
plt.xlabel("Fare")

plt.show()
```

```python
plt.figure(figsize=(10, 7))

correlation = df.select_dtypes(include="number").corr()

sns.heatmap(correlation, annot=True, cmap="coolwarm", fmt=".2f")

plt.title("Correlation Heatmap of Titanic Numerical Variables")

plt.show()
```

```python
plt.figure(figsize=(7, 5))

sns.boxplot(x="Survived", y="Age", data=df)

plt.title("Age Distribution by Survival Status")
plt.xlabel("Survived (0 = No, 1 = Yes)")
plt.ylabel("Age")

plt.show()
```

```python
plt.figure(figsize=(7, 5))

sns.boxplot(x="Pclass", y="Fare", data=df)

plt.title("Fare Distribution by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Fare")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.barplot(x="Pclass", y="Survived", hue="Sex", data=df)

plt.title("Survival Rate by Passenger Class and Gender")
plt.xlabel("Passenger Class")
plt.ylabel("Survival Rate")
plt.legend(title="Gender")

plt.show()
```

```python
total_passengers = len(df)
survived = df["Survived"].sum()
not_survived = total_passengers - survived

print("Total Passengers:", total_passengers)
print("Survived:", survived)
print("Did Not Survive:", not_survived)
print("Overall Survival Rate:", round((survived / total_passengers) * 100, 2), "%")
```

## Top 5 Analytical Insights

1. **Overall Survival:** Out of 891 passengers, 342 survived, giving an overall survival rate of 38.38%.

2. **Gender and Survival:** Female passengers had a substantially higher survival rate (74.2%) than male passengers (18.9%).

3. **Passenger Class and Survival:** Survival rate decreased with passenger class. First-class passengers had the highest survival rate (63.0%), while third-class passengers had the lowest (24.2%).

4. **Fare Distribution:** Fare values were right-skewed, with most passengers paying relatively low fares and a small number paying very high fares.

5. **Gender and Class:** Female passengers had higher survival rates than male passengers across all three passenger classes.

## Conclusion

The Exploratory Data Analysis of the Titanic dataset provided useful insights into the factors associated with passenger survival. The analysis showed that gender and passenger class had a strong relationship with survival outcomes. Female passengers had a much higher survival rate than male passengers, while first-class passengers had a higher survival rate than second- and third-class passengers.

The analysis also showed that fare values were right-skewed, with most passengers paying lower fares and a small number paying very high fares. Overall, 342 out of 891 passengers survived, resulting in an overall survival rate of 38.38%.

These findings demonstrate how data analysis and visualization can be used to identify patterns and relationships within a dataset and provide meaningful insights from raw data.
