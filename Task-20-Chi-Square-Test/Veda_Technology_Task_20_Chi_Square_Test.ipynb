```python
import pandas as pd
import numpy as np
from scipy.stats import chi2_contingency

print("Libraries imported successfully!")
```

```python

data = pd.DataFrame({
    "Gender": [
        "Male","Male","Male","Male","Male","Male","Male","Male","Male","Male",
        "Female","Female","Female","Female","Female","Female","Female","Female","Female","Female"
    ],
    "Survived": [
        "No","No","Yes","No","Yes","No","No","Yes","No","No",
        "Yes","Yes","No","Yes","Yes","Yes","No","Yes","Yes","Yes"
    ]
})

print("Dataset Shape:", data.shape)
data.head()
```

```python
contingency_table = pd.crosstab(
    data["Gender"],
    data["Survived"]
)

print("Contingency Table:")
contingency_table
```

```python
chi2, p_value, dof, expected = chi2_contingency(contingency_table)

print("Chi-Square Test of Independence")
print("--------------------------------")
print("Chi-square statistic:", round(chi2, 4))
print("Degrees of freedom:", dof)
print("P-value:", round(p_value, 4))
```

```python
expected_table = pd.DataFrame(
    expected,
    index=contingency_table.index,
    columns=contingency_table.columns
)

print("Expected Frequencies:")
expected_table
```

```python
alpha = 0.05

print("Significance Level:", alpha)

if p_value < alpha:
    print("\nConclusion: Reject H0.")
    print("There is a statistically significant association between Gender and Survival.")
else:
    print("\nConclusion: Fail to reject H0.")
    print("There is not enough evidence to conclude that Gender and Survival are associated.")
```

Interpretation

A Chi-Square Test of Independence was performed to determine whether two categorical variables, Gender and Survival, were statistically associated.

Hypotheses

Null Hypothesis (H0): Gender and Survival are statistically independent.

Alternative Hypothesis (H1): Gender and Survival are statistically associated.

A significance level of 0.05 was used.

The contingency table showed the observed frequencies for each combination of Gender and Survival. Expected frequencies were then calculated under the assumption that the two variables were independent.

The calculated Chi-Square statistic and p-value were used to make the final decision.

If the p-value is less than 0.05, the null hypothesis is rejected, indicating a statistically significant association between the variables.

If the p-value is greater than or equal to 0.05, the null hypothesis is not rejected because there is insufficient evidence to conclude that the variables are associated.

Conclusion

The Chi-Square Test of Independence provides a useful statistical method for studying relationships between categorical variables. The analysis included creation of a contingency table, calculation of expected frequencies, execution of the Chi-Square test, and interpretation of the p-value.
