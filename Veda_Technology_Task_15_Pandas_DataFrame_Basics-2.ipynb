```python
from sklearn.datasets import load_iris

iris = load_iris()

iris_df = pd.DataFrame(
    iris.data,
    columns=iris.feature_names
)

iris_df["species"] = [
    iris.target_names[i] for i in iris.target
]

iris_df.head()
```

## Interview Questions

### 1. What is a Pandas DataFrame?

A Pandas DataFrame is a two-dimensional labeled data structure in Python. It stores data in rows and columns, similar to a table or spreadsheet.

### 2. What information does `info()` provide?

The `info()` function provides information about the DataFrame structure, including the number of rows, column names, non-null values, data types, and memory usage.

### 3. What is the difference between `shape` and `size`?

`shape` returns the number of rows and columns in a DataFrame.

For example:

`(150, 5)` means 150 rows and 5 columns.

`size` returns the total number of elements in the DataFrame.

For a DataFrame with 150 rows and 5 columns:

`150 × 5 = 750` elements.
