```python
import pandas as pd
import numpy as np
from scipy import stats

print("Libraries imported successfully!")
```

```python
np.random.seed(42)

# Sample data
data = pd.DataFrame({
    "Group_A": np.random.normal(70, 10, 30),
    "Group_B": np.random.normal(75, 10, 30),
    "Scores": np.random.normal(72, 8, 30)
})

data.head()
```

```python
alpha = 0.05
print("Significance level:", alpha)
```

```python
sample = data["Scores"]

t_stat, p_value = stats.ttest_1samp(sample, 70)

print("Test 1: One-Sample t-test")
print("Null Hypothesis (H0): Mean score = 70")
print("Alternative Hypothesis (H1): Mean score != 70")
print("Test Statistic:", round(t_stat, 4))
print("P-value:", round(p_value, 4))

if p_value < alpha:
    print("Conclusion: Reject H0. The mean score is significantly different from 70.")
else:
    print("Conclusion: Fail to reject H0. There is not enough evidence that the mean score differs from 70.")
```

```python
t_stat, p_value = stats.ttest_ind(
    data["Group_A"],
    data["Group_B"]
)

print("Test 2: Independent Two-Sample t-test")
print("Null Hypothesis (H0): Mean of Group A = Mean of Group B")
print("Alternative Hypothesis (H1): Mean of Group A != Mean of Group B")
print("Test Statistic:", round(t_stat, 4))
print("P-value:", round(p_value, 4))

if p_value < alpha:
    print("Conclusion: Reject H0. The group means are significantly different.")
else:
    print("Conclusion: Fail to reject H0. There is not enough evidence that the group means are different.")
```

```python
observed = np.array([
    [30, 20],
    [20, 30]
])

chi2, p_value, dof, expected = stats.chi2_contingency(observed)

print("Test 3: Chi-Square Test")
print("Null Hypothesis (H0): The two categorical variables are independent.")
print("Alternative Hypothesis (H1): The two categorical variables are associated.")
print("Chi-Square Statistic:", round(chi2, 4))
print("P-value:", round(p_value, 4))

if p_value < alpha:
    print("Conclusion: Reject H0. There is a significant association.")
else:
    print("Conclusion: Fail to reject H0. There is not enough evidence of an association.")
```

# Hypothesis Testing – Interpretation Report

## Objective
The objective of this task was to understand the fundamentals of statistical hypothesis testing by formulating null and alternative hypotheses and performing basic statistical tests.

## Significance Level
A significance level of 0.05 was used for all tests.

## Tests Performed

### 1. One-Sample t-test
The test was used to determine whether the sample mean score was significantly different from a reference value of 70.

- H0: Mean score = 70
- H1: Mean score ≠ 70

The p-value was compared with 0.05 to determine whether the null hypothesis should be rejected.

### 2. Independent Two-Sample t-test
The test was used to compare the mean scores of Group A and Group B.

- H0: Mean of Group A = Mean of Group B
- H1: Mean of Group A ≠ Mean of Group B

The p-value was used to determine whether there was a statistically significant difference between the two group means.

### 3. Chi-Square Test
The chi-square test was used to examine whether two categorical variables were statistically independent.

- H0: The categorical variables are independent.
- H1: The categorical variables are associated.

The p-value was compared with the significance level to make the final decision.

## Conclusion
Hypothesis testing provides a systematic method for making decisions about population characteristics using sample data. The null and alternative hypotheses were formulated before performing each test, and the resulting test statistics and p-values were used to make conclusions.

A p-value below 0.05 provides sufficient evidence to reject the null hypothesis, while a p-value equal to or above 0.05 means that there is not enough evidence to reject the null hypothesis.

The task helped develop a practical understanding of statistical significance and the correct interpretation of p-values.
