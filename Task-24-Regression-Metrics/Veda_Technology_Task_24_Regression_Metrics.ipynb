```python
import pandas as pd
import numpy as np

from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

print("Libraries imported successfully!")
```

```python
actual = np.array([100, 120, 150, 180, 200, 220, 250, 280, 300, 350])

predicted = np.array([105, 115, 145, 190, 195, 225, 240, 290, 310, 340])

print("Actual values:")
print(actual)

print("\nPredicted values:")
print(predicted)
```

```python
mae = mean_absolute_error(actual, predicted)
mse = mean_squared_error(actual, predicted)
rmse = np.sqrt(mse)
r2 = r2_score(actual, predicted)

print("Regression Metrics")
print("------------------")
print("MAE  :", round(mae, 4))
print("MSE  :", round(mse, 4))
print("RMSE :", round(rmse, 4))
print("R²   :", round(r2, 4))
```

```python
comparison = pd.DataFrame({
    "Metric": ["MAE", "MSE", "RMSE", "R²"],
    "Value": [mae, mse, rmse, r2],
    "Interpretation": [
        "Average absolute prediction error",
        "Average squared prediction error",
        "Typical prediction error with more penalty for large errors",
        "Proportion of target variance explained by the model"
    ]
})

comparison
```

```python
errors = actual - predicted

error_table = pd.DataFrame({
    "Actual": actual,
    "Predicted": predicted,
    "Error": errors,
    "Absolute Error": np.abs(errors)
})

error_table
```

## Interpretation of Regression Metrics

### MAE – Mean Absolute Error

MAE represents the average absolute difference between actual and predicted values.

It is easy to understand because it is expressed in the same units as the target variable.

Lower MAE indicates smaller average prediction errors.

### MSE – Mean Squared Error

MSE calculates the average of squared prediction errors.

Because errors are squared, larger errors receive greater penalty.

Lower MSE indicates better prediction performance.

### RMSE – Root Mean Squared Error

RMSE is the square root of MSE.

It is expressed in the same units as the target variable and is more sensitive to large errors than MAE.

Lower RMSE indicates better prediction performance.

### R² – R-Squared

R² measures the proportion of variation in the target variable that is explained by the regression model.

A higher R² generally indicates that the model explains more variation in the target data.

However, R² should not be used alone to judge model performance.

## Most Appropriate Metric

There is no single regression metric that is best for every situation.

MAE is useful when the average prediction error needs to be easy to interpret and all errors should have similar importance.

RMSE is useful when larger prediction errors are more important because it penalizes large errors more strongly.

MSE is useful for mathematical optimization and model training, but its squared units can make interpretation less intuitive.

R² is useful for understanding how much variation in the target is explained by the model, but it should be considered together with error-based metrics.

For practical prediction tasks, MAE or RMSE can be selected depending on whether large errors need extra attention.

## Conclusion

The four important regression evaluation metrics, MAE, MSE, RMSE, and R², were calculated and compared.

The task demonstrated how different metrics measure regression performance in different ways. MAE provides an easily interpretable average error, MSE penalizes large errors, RMSE expresses error in the original target units while emphasizing larger errors, and R² measures the proportion of variance explained by the model.

Using multiple metrics provides a more complete understanding of regression model performance.
