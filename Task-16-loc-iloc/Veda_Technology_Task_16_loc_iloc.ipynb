```python
import pandas as pd

print("Pandas imported successfully!")
```

```python
data = {
    "Name": ["Srujan", "Vaishu", "JanAvi", "Leesha", "Nambratha"],
    "Age": [22, 21, 22, 20, 23],
    "Marks": [85, 78, 92, 88, 75],
    "Department": ["ISE", "CSE-AIML", "ISE", "CSE", "CS"]
}

df = pd.DataFrame(data, index=["S1", "S2", "S3", "S4", "S5"])

df
```

```python
print("LOC Example 1:")
df.loc["S1"]
```

```python
print("LOC Example 2:")
df.loc[["S1", "S3", "S5"]]
```

```python
print("LOC Example 3:")
df.loc["S2", "Marks"]
```

```python
print("LOC Example 4:")
df.loc[["S1", "S3"], ["Name", "Marks"]]
```

```python
print("LOC Example 5:")
df.loc[df["Marks"] > 80, ["Name", "Marks"]]
```

```python
print("ILOC Example 1:")
df.iloc[0]
```

```python
print("ILOC Example 2:")
df.iloc[[0, 2, 4]]
```

```python
print("ILOC Example 3:")
df.iloc[1, 2]
```

```python
print("ILOC Example 4:")
df.iloc[[0, 2], [0, 2]]
```

```python
print("ILOC Example 5:")
df.iloc[1:4, 0:3]
```

## Interview Questions

### 1. What is the difference between loc[] and iloc[]?

`loc[]` performs label-based selection, while `iloc[]` performs position-based selection using integer indexes.

Example:

`df.loc["S1"]` selects the row with label S1.

`df.iloc[0]` selects the first row based on its position.

### 2. When should you use loc[]?

Use `loc[]` when you want to select data using row labels, column names, or conditions.

### 3. Can iloc[] use column names?

No. `iloc[]` uses integer positions only.

For example:

`df.iloc[0, 2]`

selects the first row and third column.

To select using column names, use `loc[]`.

## Observations

`loc[]` and `iloc[]` are important Pandas DataFrame selection methods.

- `loc[]` uses labels.
- `iloc[]` uses integer positions.
- `loc[]` can use column names.
- `iloc[]` cannot directly use column names.
- Both methods can be used to select rows, columns, or specific values.

The examples demonstrated different ways to select data using both methods.
