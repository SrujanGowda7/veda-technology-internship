```python
# TASK 31 - K-MEANS CLUSTERING
# Data Science Internship - Veda Technology

# 1. Import Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

# 2. Load Dataset
iris = load_iris()

df = pd.DataFrame(
    iris.data,
    columns=iris.feature_names
)

print("First 5 rows:")
print(df.head())

print("\nDataset Shape:", df.shape)

# 3. Select Numerical Features
X = df.copy()

# 4. Scale the Features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# 5. Find Suitable Number of Clusters using Elbow Method
inertia = []

for k in range(1, 11):
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(X_scaled)
    inertia.append(kmeans.inertia_)

plt.figure(figsize=(8, 5))
plt.plot(range(1, 11), inertia, marker='o')
plt.xlabel("Number of Clusters (K)")
plt.ylabel("Inertia")
plt.title("Elbow Method")
plt.show()

# 6. Build K-Means Model
kmeans = KMeans(
    n_clusters=3,
    random_state=42,
    n_init=10
)

clusters = kmeans.fit_predict(X_scaled)

# 7. Add Cluster Labels to Dataset
df["Cluster"] = clusters

print("\nClustered Dataset:")
print(df.head(10))

# 8. Cluster Sizes
print("\nCluster Sizes:")
print(df["Cluster"].value_counts().sort_index())

# 9. Silhouette Score
score = silhouette_score(X_scaled, clusters)

print("\nSilhouette Score:", round(score, 3))

# 10. Cluster Centers
centers_scaled = kmeans.cluster_centers_

centers = scaler.inverse_transform(centers_scaled)

cluster_centers = pd.DataFrame(
    centers,
    columns=df.columns[:-1]
)

print("\nCluster Centers:")
print(cluster_centers)

# 11. Visualize Clusters
plt.figure(figsize=(8, 6))

plt.scatter(
    X_scaled[:, 0],
    X_scaled[:, 1],
    c=clusters,
    s=60
)

plt.scatter(
    centers_scaled[:, 0],
    centers_scaled[:, 1],
    marker="X",
    s=200,
    label="Cluster Centers"
)

plt.xlabel("Sepal Length (Scaled)")
plt.ylabel("Sepal Width (Scaled)")
plt.title("K-Means Clustering")
plt.legend()
plt.show()

# 12. Describe Each Cluster
print("\nCluster Description:")

for cluster in sorted(df["Cluster"].unique()):
    cluster_data = df[df["Cluster"] == cluster]

    print(f"\nCluster {cluster}:")
    print("Number of observations:", len(cluster_data))
    print("Average values:")
    print(cluster_data.drop(columns="Cluster").mean().round(2))
```
