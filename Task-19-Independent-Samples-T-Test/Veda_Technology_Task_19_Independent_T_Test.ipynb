```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

print("Libraries imported successfully!")
```

```python
np.random.seed(42)

group_a = np.random.normal(70, 10, 30)
group_b = np.random.normal(75, 10, 30)

df = pd.DataFrame({
    "Group_A": group_a,
    "Group_B": group_b
})

df.head()
```

```python
print("Group A Mean:", round(df["Group_A"].mean(), 2))
print("Group B Mean:", round(df["Group_B"].mean(), 2))

print("\nGroup A Standard Deviation:", round(df["Group_A"].std(), 2))
print("Group B Standard Deviation:", round(df["Group_B"].std(), 2))
```

```python
plt.figure(figsize=(8, 5))

plt.hist(df["Group_A"], alpha=0.6, label="Group A")
plt.hist(df["Group_B"], alpha=0.6, label="Group B")

plt.title("Distribution of Group A and Group B")
plt.xlabel("Score")
plt.ylabel("Frequency")
plt.legend()
plt.show()
```

```python
t_stat, p_value = stats.ttest_ind(
    df["Group_A"],
    df["Group_B"]
)

alpha = 0.05

print("Independent Samples t-Test")
print("----------------------------")
print("Null Hypothesis (H0): Mean of Group A = Mean of Group B")
print("Alternative Hypothesis (H1): Mean of Group A != Mean of Group B")
print()
print("t-statistic:", round(t_stat, 4))
print("p-value:", round(p_value, 4))
print("Significance Level:", alpha)

if p_value < alpha:
    print("\nConclusion: Reject H0.")
    print("There is a statistically significant difference between the two group means.")
else:
    print("\nConclusion: Fail to reject H0.")
    print("There is not enough evidence to conclude that the two group means are different.")
```

## Interpretation

An independent samples t-test was performed to compare the average values of two independent groups.

The null hypothesis stated that the means of Group A and Group B were equal, while the alternative hypothesis stated that their means were different.

A significance level of 0.05 was used. The calculated t-statistic and p-value were used to determine whether the difference between the group means was statistically significant.

If the p-value is less than 0.05, the null hypothesis is rejected and the difference between the group means is considered statistically significant.

If the p-value is greater than or equal to 0.05, the null hypothesis is not rejected because there is insufficient evidence to conclude that the group means are significantly different.

The distribution plot was also inspected before performing the test to understand the values in both groups.

## Conclusion

The independent samples t-test provides a statistical method for determining whether two independent groups have significantly different average values. It is useful when comparing the means of two separate groups, such as the performance of two classes or scores of two different groups of students.
