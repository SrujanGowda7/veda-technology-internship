# Task 4 – Data Visualization with Matplotlib & Seaborn

## Objective
Create meaningful visualizations to understand patterns and relationships in the Titanic dataset using Matplotlib and Seaborn.

## Tools
- Python
- Pandas
- NumPy
- Matplotlib
- Seaborn

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

print("Libraries imported successfully!")
```

```python
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"

df = pd.read_csv(url)

print("Dataset Shape:", df.shape)
df.head()
```

```python
df.info()
```

## Visualization 1 – Survival Distribution

**Question:** How many passengers survived and how many did not?

**Chart Selection:** A bar chart is suitable for comparing the number of passengers across two categories.

```python
survival_counts = df["Survived"].value_counts().sort_index()

plt.figure(figsize=(7, 5))
plt.bar(["Did Not Survive", "Survived"], survival_counts.values)
plt.title("Titanic Passenger Survival Distribution")
plt.xlabel("Survival Status")
plt.ylabel("Number of Passengers")
plt.show()

print("Takeaway: More passengers did not survive than survived the Titanic disaster.")
```

## Visualization 2 – Passenger Class Distribution

**Question:** How were passengers distributed across the three passenger classes?

**Chart Selection:** A count plot is useful for comparing the number of observations in different categorical groups.

```python
plt.figure(figsize=(7, 5))
sns.countplot(data=df, x="Pclass")
plt.title("Passenger Distribution by Class")
plt.xlabel("Passenger Class")
plt.ylabel("Number of Passengers")
plt.show()

print("Takeaway: Third-class passengers formed the largest passenger group.")
```

## Visualization 3 – Age Distribution

**Question:** What is the age distribution of Titanic passengers?

**Chart Selection:** A histogram is appropriate for showing the distribution of a continuous numerical variable such as age.

```python
plt.figure(figsize=(8, 5))
plt.hist(df["Age"].dropna(), bins=20)
plt.title("Age Distribution of Titanic Passengers")
plt.xlabel("Age")
plt.ylabel("Number of Passengers")
plt.show()

print("Takeaway: Most passengers were concentrated in the young-adult to middle-age range.")
```

## Visualization 4 – Fare Distribution by Passenger Class

**Question:** How does the fare vary across passenger classes?

**Chart Selection:** A boxplot shows the median, spread, and possible outliers of fares for each passenger class.

```python
plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x="Pclass", y="Fare")
plt.title("Fare Distribution by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Fare")
plt.show()

print("Takeaway: First-class passengers generally paid substantially higher fares than second- and third-class passengers.")
```

## Visualization 5 – Age vs Fare

**Question:** Is there a relationship between passenger age and fare?

**Chart Selection:** A scatter plot is useful for examining the relationship between two numerical variables.

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x="Age", y="Fare", alpha=0.6)
plt.title("Relationship Between Age and Fare")
plt.xlabel("Age")
plt.ylabel("Fare")
plt.show()

print("Takeaway: Age and fare do not show a strong direct relationship, although some high-fare outliers are visible.")
```

## Visualization 6 – Correlation Heatmap

**Question:** What relationships exist between the numerical variables?

**Chart Selection:** A heatmap provides a compact visual representation of correlations between multiple numerical variables.

```python
numeric_df = df.select_dtypes(include=np.number)

plt.figure(figsize=(10, 7))
sns.heatmap(numeric_df.corr(), annot=True, fmt=".2f", cmap="coolwarm")
plt.title("Correlation Heatmap of Titanic Numerical Variables")
plt.show()

print("Takeaway: Passenger class and fare show a noticeable relationship, while several other variables have weaker correlations.")
```

## Visualization 7 – Survival by Gender

**Question:** How did survival differ between male and female passengers?

**Chart Selection:** A grouped count plot makes it easy to compare survival outcomes across gender categories.

```python
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x="Sex", hue="Survived")
plt.title("Survival Distribution by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Passengers")
plt.legend(title="Survived", labels=["No", "Yes"])
plt.show()

print("Takeaway: Female passengers had a much higher survival proportion than male passengers.")
```

# Overall Findings

- The number of passengers who did not survive was higher than the number who survived.
- Third-class passengers represented the largest passenger group.
- Passenger ages were mainly concentrated among young and middle-aged adults.
- First-class passengers generally paid higher fares.
- Age and fare did not have a strong direct relationship.
- Passenger class and fare showed a noticeable relationship.
- Female passengers had a substantially higher survival proportion than male passengers.

## Conclusion

The visualizations reveal important patterns in the Titanic dataset and demonstrate how different chart types can be selected according to different analytical questions.
