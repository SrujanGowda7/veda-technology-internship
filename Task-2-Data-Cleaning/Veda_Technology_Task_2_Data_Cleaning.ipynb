```python
import pandas as pd
import numpy as np
```

```python
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"

df = pd.read_csv(url)

print("Dataset loaded successfully!")
print("Shape:", df.shape)

df.head()
```

```python
print("Dataset Information:")
df.info()

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())
```

```python
print("Unique values in categorical columns:")

print("\nSex:")
print(df["Sex"].unique())

print("\nEmbarked:")
print(df["Embarked"].unique())

print("\nPclass:")
print(df["Pclass"].unique())
```

```python
# Check duplicate rows
duplicates_before = df.duplicated().sum()

print("Duplicate rows found:", duplicates_before)

# Remove duplicate rows
df = df.drop_duplicates()

duplicates_after = df.duplicated().sum()

print("Duplicate rows after cleaning:", duplicates_after)
```

```python
# Handle missing values

# Fill missing Age values with median
df["Age"] = df["Age"].fillna(df["Age"].median())

# Fill missing Embarked values with most frequent value
df["Embarked"] = df["Embarked"].fillna(df["Embarked"].mode()[0])

# Drop Cabin because it contains too many missing values
df = df.drop(columns=["Cabin"])

print("Missing values after cleaning:")
print(df.isnull().sum())
```

```python
# Standardize categorical values

# Remove extra spaces and standardize text format
df["Sex"] = df["Sex"].str.strip().str.lower()
df["Embarked"] = df["Embarked"].str.strip().str.upper()

# Convert Sex values to consistent labels
df["Sex"] = df["Sex"].replace({
    "male": "Male",
    "female": "Female"
})

print("Standardized Sex values:")
print(df["Sex"].unique())

print("\nStandardized Embarked values:")
print(df["Embarked"].unique())
```

```python
# Check and correct data types

df["PassengerId"] = df["PassengerId"].astype(int)
df["Survived"] = df["Survived"].astype(int)
df["Pclass"] = df["Pclass"].astype(int)
df["Age"] = df["Age"].astype(float)
df["Fare"] = df["Fare"].astype(float)

print("Data types after correction:")
print(df.dtypes)
```

```python
# Final data quality check

print("Final dataset shape:")
print(df.shape)

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate rows:")
print(df.duplicated().sum())

print("\nFinal data types:")
print(df.dtypes)

print("\nFirst 5 rows of cleaned dataset:")
display(df.head())
```

```python
# Save cleaned dataset as CSV

output_file = "titanic_cleaned.csv"

df.to_csv(output_file, index=False)

print("Cleaned dataset saved successfully!")
print("File name:", output_file)
```
