```python
import numpy as np

print("NumPy imported successfully!")
```

```python
random_integers = np.random.randint(1, 101, size=10)

print("Random Integers:")
print(random_integers)
```

```python
random_floats = np.random.random(10)

print("Random Floating-Point Numbers:")
print(random_floats)
```

```python
students = np.array([
    "Student_1", "Student_2", "Student_3", "Student_4", "Student_5",
    "Student_6", "Student_7", "Student_8", "Student_9", "Student_10"
])

sample = np.random.choice(students, size=5, replace=False)

print("Original Array:")
print(students)

print("\nRandom Sample:")
print(sample)
```

```python
np.random.seed(42)

marks = np.random.randint(40, 101, size=10)

student_data = {
    "Student": np.arange(1, 11),
    "Marks": marks
}

print("Simulated Student Marks:")
print(student_data)
```

```python
np.random.seed(42)
result_1 = np.random.randint(1, 101, size=5)

np.random.seed(42)
result_2 = np.random.randint(1, 101, size=5)

print("First Result:")
print(result_1)

print("\nSecond Result:")
print(result_2)

print("\nAre both results identical?", np.array_equal(result_1, result_2))
```

## Interpretation

NumPy provides several functions for generating random data, including random integers, floating-point numbers, and random samples from arrays.

A random seed was used to make the generated results reproducible. When the same seed is used with the same random-generation process, the same sequence of values is produced.

Random sampling is useful in data science for simulations, experimentation, sampling datasets, testing algorithms, and creating synthetic data.

Reproducibility is important because it allows other researchers or developers to obtain the same results and verify an experiment.

## Interview Questions

### 1. Why are random numbers useful in data science?

Random numbers are useful for sampling, simulations, experiments, creating synthetic datasets, and testing machine learning algorithms.

### 2. What is a random seed?

A random seed is an initial value used by a random number generator. Using the same seed produces the same sequence of random values.

### 3. Why is reproducibility important?

Reproducibility ensures that an experiment can be repeated and produce the same results. It helps with debugging, verification, research, and reliable data science experiments.
